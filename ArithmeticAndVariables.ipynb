```python

# Lets calculate number of seconds in two weeks

num_of_weeks = 2
num_of_days_in_one_week = 7
num_of_hours_in_a_day = 24
num_of_min_in_a_hour = 60
num_of_sec_in_a_min = 60

num_of_sec_in_two_weeks = num_of_weeks * num_of_days_in_one_week * num_of_hours_in_a_day * num_of_min_in_a_hour * num_of_sec_in_a_min

print(num_of_sec_in_two_weeks);
```

```python
# Lets calculate number of seconds in two years

num_of_years = 2
num_of_week_in_a_year = 52
num_of_days_in_one_week = 7
num_of_hours_in_a_day = 24
num_of_min_in_a_hour = 60
num_of_sec_in_a_min = 60

num_of_sec_in_two_years = num_of_years * num_of_week_in_a_year * num_of_days_in_one_week * num_of_hours_in_a_day * num_of_min_in_a_hour * num_of_sec_in_a_min

print(num_of_sec_in_two_years)
```

```python
# Update to include leap years

days_per_year = 365.25
num_of_hours_in_a_day = 24
num_of_min_in_a_hour = 60
num_of_sec_in_a_min = 60

num_of_seconds_in_leap_year = days_per_year * num_of_hours_in_a_day * num_of_min_in_a_hour * num_of_sec_in_a_min

print(num_of_seconds_in_leap_year)
```
